```python
from sampiclyser import SAMPIC_Run_Decoder
from natsort import natsorted
from pathlib import Path
```

```python
import os
```

```python
ls_data_dir = os.listdir('./data')
```

```python
ls_data_dir
```

```python
#prefix = []
daterun = []
runno = []

for item in ls_data_dir:
    a=item.split('_')
    if len(a)>1:
        #prefix.append(a[0])
        daterun.append(a[1]+"_"+a[2])
        runno.append(a[3])

    
```

```python
outputdir= "PPS2_LGAD_Testbeam_"+daterun[0][:6]
try :
    print("Creating directory :",outputdir)
    os.mkdir(outputdir)
except FileExistsError:
    print("Directory ",outputdir," already exists for this campaign")
    
```

```python
runno
```

```python
base_path = Path("./data")
```

```python
for i,run in enumerate(runno):
    j=i+1
    run_path = base_path/ls_data_dir[j]
    print(run_path)
    run_files = natsorted(list(run_path.glob("*.bin*")))
    print(run_files)
    _decoder = SAMPIC_Run_Decoder(run_path)
    outputfile_parquet=Path("./"+outputdir+"/"+run+".parquet")
    outputfile_root=Path("./"+outputdir+"/"+run+".root")
    #    outputfile_parquet=Path("./"+outputdir+"/"+runno[i]+".parquet")
    print(outputfile_parquet ,"\n", outputfile_root)
    _decoder.decode_data(root_path=outputfile_root,parquet_path=outputfile_parquet)

    
    
#decoder.decode_data(feather_path=Path("./Run001.feather"))
#decoder.decode_data(root_path=Path("./Run001.root"))
# It is possible to decode simultaneously to multiple formats:
#decoder.decode_data(parquet_path=Path("./Run001.parquet"), feather_path=Path("./Run001.feather"), root_path=Path("./Run001.root"))
```
